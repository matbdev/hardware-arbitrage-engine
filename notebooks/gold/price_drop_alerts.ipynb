3. gold_price_drop_alerts (Monitoramento de Negociações)

Criada especificamente para identificar vendedores que estão "desesperados" e abaixando o preço aos poucos porque não conseguem vender.

    Granularidade: 1 linha = 1 anúncio que sofreu alteração negativa de preço.

    Colunas Sugeridas:

        ad_id (PK)

        title, url

        initial_price (Preço quando o anúncio foi coletado pela primeira vez).

        current_price (Preço da última coleta).

        price_drop_pct (Porcentagem de queda de valor).

        days_on_market (Dias desde a primeira coleta - anúncios antigos aceitam propostas mais agressivas).
