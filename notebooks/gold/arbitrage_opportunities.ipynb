Esta é a tabela que vai alimentar a tela inicial do seu sistema. Ela contém apenas os anúncios ativos que possuem margem de lucro matemática comprovada, ordenados do melhor para o pior negócio.

    Granularidade: 1 linha = 1 anúncio ativo viável.

    Colunas Sugeridas:

        ad_id (PK)

        title, url, first_image_src, item_condition

        specs_summary (Ex: "Intel Core i5, 16GB RAM, 512GB SSD" - string concatenada para não poluir o frontend).

        current_price (O preço listado no anúncio).

        market_median_price (O preço de mercado daquele exato modelo/configuração).

        potential_profit (Lucro bruto projetado em R$).

        profit_margin_pct (Margem em porcentagem).

        is_urgent_sale (Booleano derivado da Silver, excelente para negociação).

        opportunity_score (Um ranking interno de 0 a 100 baseado em margem, urgência e condição).
