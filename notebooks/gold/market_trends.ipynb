4. gold_market_trends (Para Gráficos e Analytics)

Se o seu frontend tiver uma aba de estatísticas, essa tabela agregada alimenta gráficos temporais sem precisar consultar a Silver inteira.

    Granularidade: 1 linha = 1 dia + 1 Categoria/Configuração.

    Colunas Sugeridas:

        date (Data da consolidação).

        category, brand (Ex: Notebooks Samsung).

        average_market_price (Evolução do preço no dia).

        total_volume_available (Quantidade de itens à venda no mercado).
