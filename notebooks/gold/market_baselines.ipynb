2. gold_market_baselines (A Tabela FIPE do Hardware)

Esta tabela serve como um dicionário de preços para a API e pode ser usada em uma tela de consulta manual (onde você digita as configurações e ele retorna o preço base).

    Granularidade: 1 linha = 1 agrupamento único de especificações (ex: Categoria + CPU + RAM + Armazenamento).

    Colunas Sugeridas:

        baseline_id (PK gerada por hash das características).

        category, cpu_brand, ram_gb, storage_gb

        active_ads_count (Quantos anúncios no mercado sustentam essa métrica - para evitar confiar na mediana de 1 anúncio só).

        min_price (Preço mínimo histórico).

        median_price (O preço "justo" a ser considerado).

        max_price (Preço teto).

        last_recalculated_at (Data da última atualização).
